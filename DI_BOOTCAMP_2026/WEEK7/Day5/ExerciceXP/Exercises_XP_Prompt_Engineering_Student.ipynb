## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = """As an educational content creator for 7th graders, your task is to generate a summary and a quiz about volcanic eruptions. Please follow these instructions carefully:

SUMMARY:
- Provide a 2-bullet point summary of volcanic eruptions. Each bullet should be concise and easy to understand for an 11-13 year old.

QUIZ:
- Create 3 multiple-choice questions about volcanic eruptions.
- For each question, provide 1 correct answer and 2 plausible but incorrect distractors.
- Ensure the language is friendly and uses simple vocabulary appropriate for 7th graders.

FORMATTING:
- Present the output in a format suitable for Google Slides, clearly separating the summary and each quiz question with its options.

This prompt is designed to be reusable for creating educational content for various topics for 7th graders."""
quiz_prompt
```

```python
improvements = [
    "**Clearer instructions:** Instead of 'Make a quiz', it specifies '2-bullet summary + 3 MCQs (1 correct, 2 distractors)', giving precise output requirements.",
    "**Targeted audience and tone:** It explicitly defines the audience as '7th graders' and requires 'friendly tone, simple vocab', which the vague prompt completely lacks.",
    "**Output format constraint:** It requests 'Google Slides-ready output', providing a structural constraint that helps ensure the generated content is immediately usable, unlike the original.",
]
improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
chosen_style = 'Conversational'
prompt_template = """You are 'FlowBot', a friendly and helpful customer support chatbot for FlowNest, a productivity app. Your main goal is to assist users with common inquiries about features, troubleshooting, and account management, and guide them to relevant resources or escalate to human support when necessary. Always maintain a positive and empathetic tone. Encourage users to ask follow-up questions if their issue isn't fully resolved.

Instructions:
- Start by greeting the user warmly and asking how you can help.
- Respond in natural, easy-to-understand language.
- If a user asks about a feature, briefly explain it and offer to provide a link to the help center for more details.
- If troubleshooting, ask clarifying questions to diagnose the problem.
- If an issue is complex or sensitive (e.g., billing, security), politely inform the user you are escalating to a human agent and ask for their contact details.
- Always end your responses by asking 'Is there anything else I can assist you with today?'"""
key_style_features = [
    'Initiates conversation with a warm greeting and open-ended question.',
    'Encourages follow-up questions and offers further assistance, simulating a natural dialogue flow.',
]
justification = """The 'Conversational' style is best for a customer support chatbot because it fosters a more natural, human-like interaction, which can reduce user frustration and improve satisfaction. Unlike 'Functional' (too rigid), 'Exploratory' (too open-ended for support), or 'Structured' (can feel robotic), a conversational approach allows for dynamic question-and-answer exchanges, empathy, and clear guidance, making the support experience more engaging and effective. It builds rapport and encourages users to fully explain their issues without feeling like they are talking to a machine."""
chosen_style, prompt_template, key_style_features, justification
```

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.

```python
no_hallucination_prompt = """Summarize the provided article about climate change and marine ecosystems. Ensure that all information in your summary is directly and explicitly supported by the text. Do not introduce any outside information, assumptions, or inferences. If the text does not cover a point, do not mention it."""
no_hallucination_prompt
```

```python
verification_prompt = """Review the provided article on climate change and marine ecosystems. Create a summary, but for every factual claim you make, explicitly cite the sentence number or paragraph from the original text that supports it. If a statement cannot be verified directly from the text, omit it or clearly state that the information is not present in the provided source. Focus solely on information that can be fact-checked against the given article."""
verification_prompt
```

```python
mitigation_strategies = [
    'Grounding the model with specific source texts and strict instructions to only use provided information.',
    'Implementing fact-checking mechanisms, either automated or human-in-the-loop, to verify generated content.',
]
risky_domains = {
    'domain_1': {
        'name': 'Medical/Healthcare',
        'why': 'Hallucinations can lead to incorrect diagnoses, treatment plans, or health advice, directly endangering patient lives and well-being.',
    },
    'domain_2': {
        'name': 'Legal/Financial',
        'why': 'Incorrect or fabricated legal precedents, financial advice, or regulations can lead to severe legal penalties, financial losses, or misinformed decisions with significant real-world consequences.',
    },
}
mitigation_strategies, risky_domains
```
