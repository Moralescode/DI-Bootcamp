### 1.3 Load GPT for causal text generation
Reuse the snippet from the platform: declare the model name, tokenizer, and model weights.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# Code complété :
model_name = 'gpt2' 
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

if None in (model_name, tokenizer, model):
    raise ValueError("Remplissez model_name, tokenizer, et model avant de continuer.")

print(f"\nModèle '{model_name}' chargé avec succès !")
print("""
GPT-2 est un modèle de langage causal, ce qui signifie qu'il prédit le mot suivant dans une séquence.
Il a été entraîné sur un jeu de données diversifié et peut générer du texte cohérent et contextuellement pertinent.
""")
```

## 🌟 Exercise 2 · Transformer Architecture and Tokenization


TODO: Describe the tokenization process in your own words before coding.


```python
text = "L'apprentissage profond est fascinant"  # Code complété

if text is None:
    raise ValueError("Définissez la variable `text` avec une courte phrase.")

tokens = tokenizer.tokenize(text)                  
token_ids = tokenizer.convert_tokens_to_ids(tokens)  

print(f"Texte original : {text}")
print(f"Tokens : {tokens}")
print(f"Token IDs : {token_ids}")

# Code complété :
x_label = "Tokens"  
y_label = "Identifiants (Token IDs)"  
title = "Visualisation de la Tokenisation (GPT-2)"     

if None in (x_label, y_label, title):
    raise ValueError("Définissez x_label, y_label, et title avant d'afficher le graphique.")

plt.figure(figsize=(10, 4))
plt.bar(tokens, token_ids, color="skyblue")
plt.xlabel(x_label)
plt.ylabel(y_label)
plt.title(title)
plt.xticks(rotation=45)
plt.show()
```

## 🌟 Exercise 4 · Generate simple text


Create a fresh prompt, run the generator, and observe how the model extends your sentence token by token.


```python
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    device=0 if torch.cuda.is_available() else -1,
)

input_text = "L'avenir de l'intelligence artificielle va"  # Code complété

if input_text is None:
    raise ValueError("Définissez `input_text` avant de générer du texte.")

gen_kwargs = {
    "max_new_tokens": 40,   # Longueur de sortie pour rester concis
    "temperature": 0.7,     # Bon équilibre entre créativité et structure
    "top_p": 0.92,
    "do_sample": True,
}

# Code complété :
output_ids = generator(input_text, **gen_kwargs)  
output_text = output_ids[0]["generated_text"]      

print(f"Entrée (Prompt) : {input_text}")
print(f"Texte généré : {output_text}")
```

### EXPLICATION 

4. Paramétrage de la Génération (Exercice 4)
Pour que l'IA génère du texte de manière naturelle et contrôlée, nous avons configuré le dictionnaire gen_kwargs :

"max_new_tokens": 40 : Limite la réponse à 40 nouveaux mots/tokens pour éviter que le modèle GPT-2 (qui est bavard) ne tourne en boucle ou n'écrive un paragraphe infini.

"temperature": 0.7 : C'est le curseur de la "créativité". À 0.0, le modèle est trop rigide et répétitif. À 1.2, il devient chaotique et invente des mots. 0.7 est la valeur idéale pour avoir une réponse fluide, variée mais qui reste logique.

"top_p": 0.92 et "do_sample": True : Indiquent à l'IA de trier les mots les plus probables et de piocher parmi eux avec un soupçon de hasard, ce qui rend le texte beaucoup plus humain et moins "robotique".
