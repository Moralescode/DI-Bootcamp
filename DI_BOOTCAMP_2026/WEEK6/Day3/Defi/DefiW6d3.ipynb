## 1. Understanding BERT and XLM-RoBERTa & 2. Tokenizing Text

```python
from transformers import BertTokenizer, XLMRobertaTokenizer
import pandas as pd
from sklearn.model_selection import StratifiedKFold

# Initialize tokenizers
bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
xlm_roberta_tokenizer = XLMRobertaTokenizer.from_pretrained('xlm-roberta-base')

print("BERT Tokenizer initialized.")
print("XLM-RoBERTa Tokenizer initialized.")

# Example sentence for tokenization
sentence = "Hello, how are you doing today?"

print(f"\nOriginal Sentence: {sentence}")

# BERT Tokenization
print("\n--- BERT Tokenization ---")
bert_tokens = bert_tokenizer.tokenize(sentence)
print(f"Tokens: {bert_tokens}")
bert_encoded = bert_tokenizer(
    sentence,
    add_special_tokens=True,
    max_length=32,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)
print(f"Encoded (input_ids): {bert_encoded['input_ids'][0].tolist()}")
print(f"Attention Mask: {bert_encoded['attention_mask'][0].tolist()}")
print(f"Decoded: {bert_tokenizer.decode(bert_encoded['input_ids'][0])}")

# XLM-RoBERTa Tokenization
print("\n--- XLM-RoBERTa Tokenization ---")
xlm_roberta_tokens = xlm_roberta_tokenizer.tokenize(sentence)
print(f"Tokens: {xlm_roberta_tokens}")
xlm_roberta_encoded = xlm_roberta_tokenizer(
    sentence,
    add_special_tokens=True,
    max_length=32,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)
print(f"Encoded (input_ids): {xlm_roberta_encoded['input_ids'][0].tolist()}")
print(f"Attention Mask: {xlm_roberta_encoded['attention_mask'][0].tolist()}")
print(f"Decoded: {xlm_roberta_tokenizer.decode(xlm_roberta_encoded['input_ids'][0])}")

# Example of two-sentence tokenization (BERT)
sentence1 = "The quick brown fox."
sentence2 = "Jumps over the lazy dog."

print("\n--- BERT Two-Sentence Tokenization ---")
bert_two_sentences_encoded = bert_tokenizer(
    sentence1,
    sentence2,
    add_special_tokens=True,
    max_length=32,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)
print(f"Encoded (input_ids): {bert_two_sentences_encoded['input_ids'][0].tolist()}")
print(f"Decoded: {bert_tokenizer.decode(bert_two_sentences_encoded['input_ids'][0])}")
```

## 3. Preparing Input Data for the Model (Further Details)

```python
print("\n--- Special Tokens (BERT) ---")
print(bert_tokenizer.special_tokens_map)
print(f"Vocabulary Size (BERT): {bert_tokenizer.vocab_size}")

print("\n--- Special Tokens (XLM-RoBERTa) ---")
print(xlm_roberta_tokenizer.special_tokens_map)
print(f"Vocabulary Size (XLM-RoBERTa): {xlm_roberta_tokenizer.vocab_size}")

# Detailed example of padding and truncation
long_sentence = "This is a very long sentence that needs to be truncated to a maximum length for processing by the model."

print("\n--- Detailed Input Preparation (BERT) ---")
encoded_long = bert_tokenizer(
    long_sentence,
    add_special_tokens=True,
    max_length=15, # Shorter max_length to demonstrate truncation
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)
print(f"Original sentence length: {len(long_sentence.split())} words")
print(f"Encoded (input_ids) with padding/truncation: {encoded_long['input_ids'][0].tolist()}")
print(f"Attention Mask: {encoded_long['attention_mask'][0].tolist()}")
print(f"Decoded: {bert_tokenizer.decode(encoded_long['input_ids'][0])}")
```

## 5. Creating Cross-Validation Folds

```python
try:
    # Assuming 'label' or 'sentiment' is the target column
    target_column = 'label'
    if target_column not in train_df.columns:
        target_column = 'sentiment'

    if target_column not in train_df.columns:
        raise ValueError(f"Target column '{target_column}' not found in train_df. Please specify the correct target column.")

    X = train_df.index.values  # We will use indices to split the DataFrame
    y = train_df[target_column].values

    # Initialize StratifiedKFold
    kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    train_folds = []
    val_folds = []

    print("\nCreating 5-fold cross-validation splits...")
    for fold, (train_index, val_index) in enumerate(kf.split(X, y)):
        print(f"Fold {fold+1}:\n  Train indices length: {len(train_index)}\n  Validation indices length: {len(val_index)}")

        train_folds.append(train_index)
        val_folds.append(val_index)

        # Optional: Verify label distribution (for the first fold example)
        if fold == 0:
            train_labels = y[train_index]
            val_labels = y[val_index]
            print("  Train label distribution:")
            print(pd.Series(train_labels).value_counts(normalize=True))
            print("  Validation label distribution:")
            print(pd.Series(val_labels).value_counts(normalize=True))

    print("\nCross-validation folds created successfully.")

except NameError:
    print("Error: 'train_df' is not defined. Please ensure the previous cell for loading data ran successfully.")
except ValueError as e:
    print(f"Error creating folds: {e}")
```
