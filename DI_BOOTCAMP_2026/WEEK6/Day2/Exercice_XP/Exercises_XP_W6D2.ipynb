## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- Bidirectional encoder-only; MLM (and historically NSP): BERT utilizes a bidirectional encoder-only Transformer architecture. It is pre-trained using Masked Language Modeling (MLM), where random tokens are masked and the model predicts them based on context from both left and right, and historically Next Sentence Prediction (NSP) for sentence relationship understanding.
- Application: Sentiment analysis on customer reviews
- Why: BERT's bidirectional contextual understanding allows it to deeply grasp the full meaning of sentences, which is crucial for accurately classifying sentiment where context from both sides of a word can alter its meaning.

### GPT
- Decoder-only; autoregressive causal LM: GPT models are built on a decoder-only Transformer architecture. They are autoregressive, meaning they generate text token by token, predicting the next word based solely on the preceding words (Causal Language Modeling).
- Application: Creative content generation (e.g., writing marketing copy, poems)
- Why: Its autoregressive and generative nature makes it ideal for producing long, coherent, and creative text sequences, building upon previous tokens to create new content.

### T5
- Encoder-decoder; text-to-text framework: T5 employs an encoder-decoder Transformer architecture and frames all NLP tasks as a "text-to-text" problem, converting input text to output text, regardless of the specific task (e.g., summarization, translation).
- Application: Machine Translation (e.g., English to French)
- Why: The encoder-decoder structure is inherently suited for sequence-to-sequence tasks like translation, where an input sequence needs to be transformed into a different output sequence. Its text-to-text approach provides a unified framework for such tasks.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Improved generalization: Pre-trained models learn rich, general-purpose linguistic representations from vast amounts of data, allowing them to perform well on new, unseen tasks with minimal adaptation.
  2) Less labeled data: By leveraging knowledge gained during pre-training, these models require significantly smaller labeled datasets for fine-tuning on downstream tasks, reducing annotation costs and effort.
  3) Faster fine-tuning: Pre-trained models provide an excellent starting point, meaning the fine-tuning process converges much faster than training a model from scratch, saving computational resources and time.
  4) Transfer learning: The core idea is to transfer learned features from a large general domain to specific, often smaller, domains. The pre-trained model acts as a powerful feature extractor, reducing the need for extensive task-specific model design.
  5) Robustness: Pre-training on diverse and large datasets exposes models to a wide variety of linguistic patterns and noise, making them more robust to variations and errors in real-world data.

- Ethical concerns: Bias (reflecting and amplifying societal biases present in training data), misinformation (generating plausible but false or misleading information), misuse (e.g., generating spam, propaganda, or harmful content), privacy (memorizing and reproducing sensitive personal information from training data).
- Mitigations: Data curation (careful selection, filtering, and balancing of training data to reduce bias), differential privacy (adding statistical noise to data to protect individual privacy while still allowing for model training), safety filters (implementing post-processing mechanisms to detect and filter out harmful or inappropriate model outputs), RLHF (Reinforcement Learning from Human Feedback to align model behavior with human values and safety guidelines), audits (regular, independent evaluations for bias, fairness, and potential harmful outputs).

## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- Describe the Q/K/V flow, softmax weighting, and value mixing: For each token in the input sequence, three vectors are derived: Query (Q), Key (K), and Value (V). The attention score between a Query token and all Key tokens is computed via a dot product. These scores are then scaled and passed through a softmax function to produce attention weights, indicating how much each token should 'attend' to other tokens. Finally, these weights are multiplied by their corresponding Value vectors and summed, resulting in an output representation for the Query token that is a weighted average of all Value vectors.
- Explain why multiple heads help (subspace projections, diverse relations): Multiple attention heads allow the Transformer model to simultaneously learn different types of relationships or focus on various aspects of the input sequence. Each head can project the Q, K, and V into a different subspace, enabling it to capture diverse relational patterns—like syntactic dependencies, semantic similarities, or even coreferences—within the same sequence, leading to a richer and more comprehensive contextual representation.
- Example sentence (different from lesson):
  - "The old, rusty car barely started on the cold morning."
  - Relation 1 (e.g., Head 1 captures adjective-noun modification): `old` attends to `car`, `rusty` attends to `car`, `cold` attends to `morning`.
  - Relation 2 (e.g., Head 2 captures verb-subject/object relation): `started` attends to `car` (what started) and `morning` (when it started).

### Pre-training Objectives
- MLM vs CLM: Masked Language Modeling (MLM), characteristic of models like BERT, predicts masked tokens within a sequence by leveraging context from both left and right directions, aiming for a deep, bidirectional understanding of the text. Causal Language Modeling (CLM), used by models like GPT, predicts the next token in a sequence based only on the preceding tokens, making it an autoregressive (unidirectional) objective primarily focused on text generation.
- When to prefer MLM vs CLM: Prefer MLM for tasks requiring a deep, bidirectional understanding of context for classification, extraction, or semantic tasks (e.g., sentiment analysis, Named Entity Recognition, Question Answering). Prefer CLM for generative tasks where the model needs to produce coherent, contextually relevant text sequentially (e.g., chatbots, story generation, summarization).
- NSP: Next Sentence Prediction (NSP) was an auxiliary pre-training objective used in early BERT versions. It involved predicting whether two segments of text followed each other in the original document. While intended to help with tasks like question answering, many modern models avoid it as it often provides limited benefit for downstream tasks and sometimes distracts from the core language understanding objectives.

### Transformer Model Selection
- Sentiment on reviews: Encoder-only (e.g., BERT, RoBERTa). Justification: Sentiment analysis requires understanding the full context of a review to determine its overall tone. Encoder-only models are excellent at generating rich contextual embeddings by processing the entire input simultaneously, making them suitable for classification tasks.
- Conversational chatbot (creative responses): Decoder-only (e.g., GPT). Justification: Chatbots that generate creative responses need to produce text sequentially in an autoregressive manner, building new tokens based on previous ones. Decoder-only models are specifically designed for this kind of generative task.
- Technical document translation (EN→ES): Encoder-Decoder (e.g., T5, BART). Justification: Translation is a classic sequence-to-sequence problem where an input sequence (source language) needs to be transformed into an output sequence (target language). Encoder-decoder architectures are inherently designed to handle this transformation.

### Positional Encoding
- Purpose: Transformers process all tokens in a sequence simultaneously without any inherent sense of order. Positional encoding adds vital information about the absolute or relative position of each token within the sequence, allowing the model to understand the sequence order, grammatical structure, and dependencies that rely on word placement.
- Example failure without positions: Without positional encoding, the sentences "The cat chased the dog" and "The dog chased the cat" would likely be processed identically by the self-attention mechanism, as the model would not differentiate between the subject and object based on their positions, leading to a complete misunderstanding of which animal is chasing which.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- T=0.2: The model will produce highly deterministic, focused, and conservative outputs. It significantly sharpens the probability distribution, making the model almost always choose the most probable next token, leading to less creative but often more coherent and factually consistent text.
- T=1.5: The model will produce highly diverse, creative, and sometimes erratic or nonsensical outputs. It flattens the probability distribution, giving more weight to less likely tokens, thereby increasing randomness and the possibility of generating unexpected or imaginative text.
- T=1.0: This is the default or "neutral" temperature. The model samples according to its original predicted probability distribution without any scaling, balancing creativity and coherence as learned during training.

### 2) Application Design
- Bedtime stories (creativity vs coherence): A good strategy would be to use a moderate to high temperature (e.g., T=0.8-1.2). This encourages imaginative narratives and unexpected twists (creativity) while still maintaining a reasonable level of narrative flow and coherence, preventing the story from becoming completely nonsensical.
- Financial report summaries (accuracy/reliability): A very low temperature (e.g., T=0.1-0.3) or even T=0 (greedy decoding). The paramount concern here is factual accuracy and reliability, minimizing any creative interpretation or hallucination. A low temperature ensures the model selects the most probable and therefore most likely correct tokens.

### 3) Temperature & Bias
- How does temperature influence bias surfacing or dampening? Give a realistic example.
- Temperature can influence how biases surface. At very low temperatures (T < 1.0), the model is highly deterministic, often selecting the most common or probable token. If the training data contains strong statistical biases (e.g., "doctor" frequently co-occurring with "he"), a low temperature will deterministically output these stereotypical biases. Conversely, very high temperatures (T > 1.0) can sometimes dampen common biases by making the model sample from a wider range of less probable tokens, potentially generating less stereotypical (though also less coherent) text. However, higher temperatures can also amplify subtle biases by randomly picking on less frequent but still biased associations. At T=1.0, the model reflects the raw biases present in its learned probability distribution.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # choose a causal LM
prompt = "The wise old owl sat on the branch and"  # write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
