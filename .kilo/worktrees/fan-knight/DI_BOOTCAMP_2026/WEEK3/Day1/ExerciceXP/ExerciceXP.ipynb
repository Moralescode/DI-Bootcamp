**Exercice 1 : Création et manipulation de tableaux**
Instructions
Créez un tableau NumPy unidimensionnel contenant les nombres de 0 à 9.

```python
import numpy as np

# Creation du tableau 0 a 9
arr = np.arange(10)
print(arr)
```

**Exercice 2 : Conversion de type et opérations sur les tableaux**

Instructions

Convertissez une liste [3.14, 2.17, 0, 1, 2] en un tableau NumPy et convertissez son type de données en entier.

```python
import numpy as np

# Conversion liste -> tableau float, puis vers entier
arr_float = np.array([3.14, 2.17, 0, 1, 2])
arr_int = arr_float.astype(int)
print(arr_int)
```

**Exercice 3 : Travailler avec des tableaux multidimensionnels**

Instructions
Créez un tableau NumPy 3x3 avec des valeurs allant de 1 à 9.

```python
import numpy as np

# Tableau 3x3 de 1 a 9
arr_3x3 = np.arange(1, 10).reshape(3, 3)
print(arr_3x3)
```

**Exercice 4 : Création d'un tableau multidimensionnel avec des nombres aléatoires**

Instructions
Créez un tableau NumPy 2D de forme (4, 5) rempli de nombres aléatoires.

```python
import numpy as np

np.random.seed(42)
arr_random = np.random.rand(4, 5)
print(arr_random.round(2))
```

 **Exercice 6 : Inversion des éléments**
Instructions
Inverser l'ordre des éléments d'un tableau NumPy 1D donné (le premier élément devient le dernier).

```python
import numpy as np

arr = np.arange(10)
# Inversion du tableau
arr_reversed = arr[::-1]
print(arr_reversed)
```

**Exercice 8 : Fonctions agrégées simples**
Instructions
Trouvez la somme et la moyenne d'un tableau 1D donné.

```python
import numpy as np

arr = np.arange(10)
total = arr.sum()
average = arr.mean()
print(f"Sum: {total}, Average: {average}")
```

**Exercice 10 : Sélection conditionnelle des valeurs**

Instructions
Extraire tous les nombres impairs d'un tableau NumPy donné.

```python
import numpy as np

arr = np.arange(10)
# Selection des nombres impairs (modulo 2 != 0)
odds = arr[arr % 2 != 0]
print(odds)
```
