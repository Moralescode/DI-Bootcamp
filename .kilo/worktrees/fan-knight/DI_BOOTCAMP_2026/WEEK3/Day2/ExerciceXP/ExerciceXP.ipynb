Exercice 1 : Opérations matricielles

Instructions
Dans cet exercice, vous travaillerez avec une matrice 3x3. Voici une brève explication des concepts :

Créez une matrice 3x3 et effectuez les opérations suivantes :

1.Calculez le déterminant.

2.Trouvez l'inverse de la matrice.

```python
import numpy as np

# 1. Création d'une matrice 3x3
# Note : On choisit des valeurs qui garantissent que la matrice est inversible 
# (son déterminant ne doit pas être égal à 0).
matrice = np.array([[2, 1, 3],
                    [0, 5, 6],
                    [1, 0, 4]])

print("--- Matrice d'origine (3x3) ---")
print(matrice)
print("\n" + "="*40 + "\n")

# 2. Calcul du Déterminant
# La fonction det() renvoie un nombre décimal (float)
determinant = np.linalg.det(matrice)

print("--- 1. Déterminant ---")
print(f"Le déterminant de la matrice est : {determinant:.2f}\n")

# 3. Calcul de l'Inverse de la matrice
# La fonction inv() calcule la matrice inverse telle que : Matrice x Matrice_Inverse = Matrice_Identité
if determinant != 0:
    matrice_inverse = np.linalg.inv(matrice)
    print("--- 2. Matrice Inverse ---")
    print(matrice_inverse)
else:
    print("La matrice n'est pas inversible car son déterminant est égal à 0.")
```

Exercice 2 : Analyse statistique
Instructions
Dans cet exercice, vous calculerez des mesures statistiques pour un ensemble de données :

Moyenne : La valeur moyenne d'un ensemble de données.
Médiane : La valeur centrale d'un ensemble de données, qu'il soit classé par ordre croissant ou décroissant.
Écart type : Mesure du degré de variation ou de dispersion dans un ensemble de valeurs.
En utilisant NumPy, générez un tableau de 50 nombres aléatoires et calculez :

1. La moyenne et la médiane.
2. L'écart type.


```python
import numpy as np

# Fixer la graine (seed) pour obtenir les mêmes valeurs aléatoires à chaque exécution
np.random.seed(42)

# Génération d'un tableau de 50 nombres aléatoires (ici, des entiers entre 1 et 100)
donnees = np.random.randint(1, 101, size=50)

# 1. Calcul de la moyenne et de la médiane
moyenne = np.mean(donnees)
mediane = np.median(donnees)

# 2. Calcul de l'écart type
ecart_type = np.std(donnees)

# Affichage des résultats arrondis à 2 décimales
print("--- Analyse Statistique NumPy ---")
print(f"Moyenne    : {moyenne:.2f}")
print(f"Médiane    : {mediane:.2f}")
print(f"Écart type : {ecart_type:.2f}")
```

Exercice 3 : Manipulation de dates

Instructions

Créez un tableau NumPy de dates pour le mois de janvier 2023. Convertissez ces dates dans un autre format (par exemple, AAAA/MM/JJ)

```python
import numpy as np

# 1. Création du tableau de dates pour tout le mois de janvier 2023
# Note : La date de fin ('2023-02-01') est exclue, la séquence s'arrête donc bien au 31 janvier.
# Le paramètre [D] indique que l'unité de l'intervalle est le Jour (Day).
dates_janvier = np.arange('2023-01-01', '2023-02-01', dtype='datetime64[D]')

# 2. Conversion au format "AAAA/MM/JJ"
# Étape A : On convertit les objets datetime64 en texte (string) au format standard ISO (AAAA-MM-JJ)
dates_texte = np.datetime_as_string(dates_janvier, unit='D')

# Étape B : On remplace les tirets '-' par des slashes '/' grâce au module vectorisé np.char
dates_formatees = np.char.replace(dates_texte, '-', '/')

# 3. Affichage des résultats (Aperçu des 5 premiers éléments)
print("--- Dates d'origine (datetime64[D]) ---")
print(dates_janvier[:5], "... [tronqué]")

print("\n--- Dates converties (Format AAAA/MM/JJ) ---")
print(dates_formatees[:5], "... [tronqué]")
```

Exercice 4 : Manipulation de données avec NumPy et Pandas
Instructions
Créez un DataFrame contenant des nombres aléatoires et effectuez les opérations suivantes :

Sélection conditionnelle des données.
Fonctions d'agrégation telles que la somme et la moyenne.

```python
import numpy as np
import pandas as pd

# Fixer la graine pour la reproductibilité
np.random.seed(24)

# 1. Création d'un DataFrame avec des nombres aléatoires
# On génère une matrice de 10 lignes et 3 colonnes avec des entiers entre 1 et 100
matrice_aleatoire = np.random.randint(1, 101, size=(10, 3))

# Transformation en DataFrame Pandas avec des noms de colonnes
df = pd.DataFrame(matrice_aleatoire, columns=["Zone_A", "Zone_B", "Zone_C"])

print("--- DataFrame Original (10 lignes x 3 colonnes) ---")
print(df)
print("\n" + "="*40 + "\n")

# 2. Sélection conditionnelle des données
# Exemple : On filtre pour garder uniquement les lignes où la 'Zone_A' est supérieure à 50
filtre_conditionnel = df[df["Zone_A"] > 50]

print("--- 1. Sélection Conditionnelle (Zone_A > 50) ---")
print(filtre_conditionnel)
print("\n" + "="*40 + "\n")

# 3. Fonctions d'agrégation (Somme et Moyenne)
somme_colonnes = df.sum()    # Calcule la somme pour chaque colonne
moyenne_colonnes = df.mean() # Calcule la moyenne pour chaque colonne

print("--- 2. Fonctions d'Agrégation ---")
print("Somme totale par zone :")
print(somme_colonnes)
print("\nMoyenne par zone :")
print(moyenne_colonnes)
```

Exercice 6 : Tests d'hypothèses de base

Instructions

Créez un jeu de données d'exemple pour tester l'efficacité d'un nouveau programme de formation sur la productivité des employés :

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
```

Étant donné un ensemble de données, formulez une hypothèse simple et testez-la à l'aide des fonctions statistiques de base de NumPy.

```python
import numpy as np

# Fixer la graine pour la reproductibilité des données aléatoires
np.random.seed(42)

# --- 1. Génération du jeu de données fourni ---
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# --- 2. Calcul du gain de productivité ---
gains = productivity_after - productivity_before

# --- 3. Calculs statistiques avec les fonctions de base NumPy ---
n = len(gains)                             # Taille de l'échantillon (30)
moyenne_gain = np.mean(gains)              # Moyenne de l'échantillon
ecart_type_gain = np.std(gains, ddof=1)    # Écart-type échantillonnal (ddof=1 pour diviser par n-1)
erreur_type = ecart_type_gain / np.sqrt(n) # Erreur type de la moyenne

# --- 4. Calcul de la statistique t de Student ---
t_stat = moyenne_gain / erreur_type

# --- 5. Affichage des métriques ---
print("--- Analyse Descriptive des Données ---")
print(f"Gain moyen de productivité observé : {moyenne_gain:.2f} points")
print(f"Écart-type des gains               : {ecart_type_gain:.2f}")
print(f"Valeur de la statistique t calculée: {t_stat:.4f}\n")

# --- 6. Prise de décision (Seuil critique) ---
# Pour un échantillon de 30 employés (soit 29 degrés de liberté) au seuil de signification de 5%,
# la valeur critique théorique de t (test unilatéral) est d'environ 1.699.
t_critique = 1.699

print("--- Conclusion du Test ---")
if t_stat > t_critique:
    print(f"Le t calculé ({t_stat:.2f}) est supérieur au t critique ({t_critique}).")
    print("Décision : On REJETTE l'hypothèse nulle (H0).")
    print("Conclusion : Le programme de formation améliore de manière significative la productivité.")
else:
    print(f"Le t calculé ({t_stat:.2f}) est inférieur ou égal au t critique ({t_critique}).")
    print("Décision : On NE PEUT PAS rejeter l'hypothèse nulle (H0).")
    print("Conclusion : L'augmentation observée n'est pas statistiquement significative.")
```

Exercice 8 : Manipulation de données de séries temporelles
Instructions
Générez des données de séries chronologiques pour l'année 2023. Démontrez le découpage pour les intervalles suivants :

De janvier à mars
Avril à juin
De juillet à septembre
d'octobre à décembre
Générez des données de séries temporelles pour une période spécifique et montrez comment segmenter ces données selon différents intervalles.



```python
import numpy as np
import pandas as pd

# 1. Génération d'une séquence de dates pour toute l'année 2023
# Le paramètre freq='D' génère un index quotidien (jour par jour)
index_dates = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")

# 2. Génération de données aléatoires associées (ex: volume de ventes quotidiennes)
np.random.seed(42)
valeurs_simulees = np.random.randint(100, 1000, size=len(index_dates))

# 3. Création de la Série Temporelle Pandas
serie_temporelle = pd.Series(valeurs_simulees, index=index_dates)

# 4. Découpage (Slicing) par intervalles (Trimestres)
janvier_mars  = serie_temporelle["2023-01":"2023-03"]
avril_juin    = serie_temporelle["2023-04":"2023-06"]
juillet_sept  = serie_temporelle["2023-07":"2023-09"]
octobre_dec   = serie_temporelle["2023-10":"2023-12"]

# 5. Affichage des vérifications pour chaque intervalle
print("--- Analyse et Segmentation Spécifique (2023) ---")
print(f"Janvier à Mars     : {len(janvier_mars)} jours | Période : {janvier_mars.index.min().date()} au {janvier_mars.index.max().date()}")
print(f"Avril à Juin       : {len(avril_juin)} jours | Période : {avril_juin.index.min().date()} au {avril_juin.index.max().date()}")
print(f"Juillet à Septembre: {len(juillet_sept)} jours | Période : {juillet_sept.index.min().date()} au {juillet_sept.index.max().date()}")
print(f"Octobre à Décembre : {len(octobre_dec)} jours | Période : {octobre_dec.index.min().date()} au {octobre_dec.index.max().date()}")

print("\n--- Aperçu des premières lignes du premier intervalle (Janvier à Mars) ---")
print(janvier_mars.head(4))
```

Exercice 10 : Visualisation de base

Instructions

Utilisez Matplotlib pour visualiser un ensemble de données simple créé avec NumPy (par exemple, un graphique linéaire de nombres aléatoires).

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Fixer la graine pour obtenir le même graphique à chaque exécution
np.random.seed(42)

# 2. Génération des données avec NumPy
# On génère 50 fluctuations aléatoires (loi normale) et on calcule leur somme cumulative
donnees = np.random.normal(loc=0, scale=1, size=50).cumsum()

# 3. Création et configuration du graphique
plt.figure(figsize=(10, 5))  # Définit la taille de la fenêtre (Largeur=10, Hauteur=5)

# Tracé de la ligne principale (avec une couleur personnalisée et une épaisseur de 2)
plt.plot(donnees, label="Tendance cumulée", color="#1f77b4", linewidth=2)

# 4. Ajout des éléments d'habillage essentiels (Titre et axes)
plt.title("Visualisation de Données Aléatoires Cumulées", fontsize=14, fontweight='bold')
plt.xlabel("Index des données (Temps / Étapes)", fontsize=11)
plt.ylabel("Valeur de la variable", fontsize=11)

# Ajout d'une grille en arrière-plan pour faciliter la lecture des points
plt.grid(True, linestyle="--", alpha=0.6)

# Affichage de la légende (basée sur le paramètre 'label' défini dans plt.plot)
plt.legend()

# 5. Affichage du rendu final sur l'écran
plt.show()
```
