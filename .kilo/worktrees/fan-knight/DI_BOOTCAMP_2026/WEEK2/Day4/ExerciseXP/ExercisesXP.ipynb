### Exercice 2 : Création d'un graphique linéaire pour la variation de température
 Objectif : Créer un graphique linéaire simple à l'aide de Matplotlib qui représente les variations de température sur une semaine 

### Tâches : Utilisez une liste de valeurs de température pour chaque jour de la semaine

### Exercice 4 :  Application du test t

#### Tâche : Effectuer un test T sur deux ensembles de données générées aléatoirement.

Exemple de code :

```python
import numpy as np
from scipy import stats

# Génération de deux ensembles de données aléatoires
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Réalisation du test t
t_stat, p_value = stats.ttest_ind(data1, data2)

print(f"Statistique t : {t_stat:.4f}")
print(f"P-value : {p_value:.4f}")

if p_value < 0.05:
    print("→ Les moyennes sont significativement différentes (p < 0,05)")
else:
    print("→ Pas de différence significative (p >= 0,05)")
```

### Exercice 5 : Standardisation et normalisation des données

#### Objectif : Appliquer la standardisation (StandardScaler) et la normalisation (MinMaxScaler) sur le jeu de données « Santé mentale des étudiants ».

#### Tâche : 
Chargez le dataset `student_mental_health.csv`, gérez les valeurs manquantes, puis appliquez StandardScaler sur l'âge et MinMaxScaler.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Chargement du dataset
df = pd.read_csv("dataset/student_mental_health.csv")

# Nettoyage : conversion de Age en numérique
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())

# Standardisation de Age
scaler_std = StandardScaler()
df["Age_Standardized"] = scaler_std.fit_transform(df[["Age"]])

# Normalisation de Age
scaler_minmax = MinMaxScaler()
df["Age_Normalized"] = scaler_minmax.fit_transform(df[["Age"]])

# Visualisations
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(df["Age"], bins=15, kde=True, ax=axes[0], color="steelblue")
axes[0].set_title("Distribution originale (Age)")

sns.histplot(df["Age_Standardized"], bins=15, kde=True, ax=axes[1], color="coral")
axes[1].set_title("Après StandardScaler")

sns.histplot(df["Age_Normalized"], bins=15, kde=True, ax=axes[2], color="seagreen")
axes[2].set_title("Après MinMaxScaler")

plt.tight_layout()
plt.show()

print(df[["Age", "Age_Standardized", "Age_Normalized"]].describe())
```

### Exercice 6 : Comprendre l'ANOVA
#### Objectif : Tester la compréhension de l'ANOVA et de son application.

##### Tâche : Trois engrais différents sont appliqués à trois groupes de plantes distincts afin de tester leur efficacité. La croissance, exprimée en centimètres, est enregistrée.

#### Ensemble de données :


##### fertilizer_1: [5, 6, 7, 6, 5]
##### fertilizer_2: [7, 8, 7, 9, 8]
##### fertilizer_3: [4, 5, 4, 3, 4]


#### Questions :

##### Effectuez un test ANOVA sur les données fournies. Quelles sont les valeurs F et p ?
##### D'après la valeur P, les engrais ont-ils des effets significativement différents sur la croissance des plantes ?
##### Expliquez ce qui se passerait si la valeur P était supérieure à 0,05.

```python
import pandas as pd
import numpy as np
from scipy import stats

# Données des engrais
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Test ANOVA
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"Valeur F : {f_stat:.4f}")
print(f"P-value : {p_value:.4f}")

if p_value < 0.05:
    print("→ Les engrais ont des effets significativement différents sur la croissance (p < 0,05).")
else:
    print("→ Pas d'effet significatif des engrais (p >= 0,05).")
```
