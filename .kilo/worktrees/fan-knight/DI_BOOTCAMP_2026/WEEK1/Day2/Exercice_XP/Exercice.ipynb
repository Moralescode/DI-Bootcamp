Exercice 2 : Cinemax n 2
Principaux sujets en Python :

Parcourir les dictionnaires
Les conditionnelles
Calculs

Instructions

Ecrivez un programme qui calcule le cout total des billets de cinema pour une famille en fonction de l'age de ses membres.

L'age des membres de la famille est stocke dans un dictionnaire.
Les regles de tarification des billets sont les suivantes :
Moins de 3 ans : gratuit
De 3 a 12 ans : 10 $
Plus de 12 ans : 15 $

Donnees familiales :

family = {"rick": 43, 'beth': 13, 'morty': 5, 'summer': 8}


Parcourez le family dictionnaire pour calculer le cout total.
Imprimez le prix du billet pour chaque membre de la famille.
Imprimez le cout total a la fin.


Prime:

Permettez a l'utilisateur de saisir les noms et ages des membres de sa famille, puis calculez le cout total du billet.

```python
# Exercice 2 : Cinemax n 2
family = {"rick": 43, 'beth': 13, 'morty': 5, 'summer': 8}

total = 0

for nom, age in family.items():
    if age < 3:
        prix = 0
    elif age <= 12:
        prix = 10
    else:
        prix = 15
    total += prix
    print(f"{nom.capitalize()} ({age} ans) : {prix} $")

print(f"\nCout total : {total} $")
```

 Exercice 3 : Zara
Principaux sujets en Python :

Creation de dictionnaires
Acceder aux elements du dictionnaire et les modifier
Les methodes de dictionnaire comme .pop() et .update()

Instructions

Creer et manipuler un dictionnaire contenant des informations sur la marque Zara.

Informations sur la marque :

name: Zara
creation_date: 1975
creator_name: Amancio Ortega Gaona
type_of_clothes: men, women, children, home
international_competitors: Gap, H&M, Benetton
number_stores: 7000
major_color: 
    France: blue, 
    Spain: red, 
    US: pink, green


Creez un dictionnaire contenant brand les donnees fournies.
Modifiez et accedez au dictionnaire comme suit :
Modifiez la valeur de number_stores a 2.
Imprimez une phrase decrivant les clients de Zara en utilisant la type_of_clothes legende.
Ajoutez une nouvelle cle country_creation avec la valeur Spain.
Verifiez si international_competitors cela existe et, si oui, ajoutez « Desigual » a la liste.
Supprimez la creation_date cle.
Imprimez le dernier element dans international_competitors.
Imprimer les principales couleurs aux Etats-Unis.
Affichez le nombre de cles dans le dictionnaire.
Imprimez toutes les cles du dictionnaire.


Prime:

Creez un autre dictionnaire appele more_on_zara avec creation_date et number_stores. Fusionnez ce dictionnaire avec le dictionnaire original brand et affichez le resultat.

```python
# Exercice 3 : Zara
brand = {
    "name": "Zara",
    "creation_date": 1975,
    "creator_name": "Amancio Ortega Gaona",
    "type_of_clothes": ["men", "women", "children", "home"],
    "international_competitors": ["Gap", "H&M", "Benetton"],
    "number_stores": 7000,
    "major_color": {
        "France": "blue",
        "Spain": "red",
        "US": ["pink", "green"]
    }
}

# Modifier number_stores a 2
brand["number_stores"] = 2

# Phrase decrivant les clients
clients = ", ".join(brand["type_of_clothes"])
print(f"Zara habille les clients suivants : {clients}.")

# Ajouter country_creation
brand["country_creation"] = "Spain"

# Ajouter Desigual si international_competitors existe
if "international_competitors" in brand:
    brand["international_competitors"].append("Desigual")

# Supprimer creation_date
brand.pop("creation_date", None)

# Dernier element de international_competitors
print("Dernier competitor :", brand["international_competitors"][-1])

# Couleurs principales aux US
print("Couleurs US :", brand["major_color"]["US"])

# Nombre de cles
print("Nombre de cles :", len(brand))

# Toutes les cles
print("Cles du dictionnaire :", list(brand.keys()))

# Prime : fusion avec more_on_zara
more_on_zara = {
    "creation_date": 1975,
    "number_stores": 7000
}

brand_fusion = brand.copy()
brand_fusion.update(more_on_zara)
print("\nDictionnaire fusionne :")
for k, v in brand_fusion.items():
    print(f"  {k} : {v}")
```

```python
# Exercice 5 : Aleatoire
import random

def comparer_nombre(nombre):
    aleatoire = random.randint(1, 100)
    if nombre == aleatoire:
        print("Success!")
    else:
        print(f"Fail! Your number: {nombre}, Random number: {aleatoire}")

comparer_nombre(42)
```

Exercice 7 : Conseils sur la temperature
Objectif : Generer une temperature aleatoire et fournir des conseils en fonction de cette plage de temperatures.

Principaux sujets en Python :

Fonctions
Conditionnelles (if/ elif)
nombres aleatoires
Nombres a virgule flottante (Bonus)
Gestion des saisons (Bonus)


Etape 1 : Creer la get_random_temp() fonction

Creez une fonction get_random_temp() qui renvoie un entier aleatoire compris entre -10 et 40 degres Celsius.


Etape 2 : Creer la main() fonction

Creez une fonction appelee main(). A l'interieur de cette fonction :
Appelez get_random_temp() pour obtenir une temperature aleatoire.
Stockez la temperature dans une variable et affichez un message convivial comme :
« La temperature actuelle est de 32 degres Celsius. »


Etape 3 : Conseils selon la temperature

Moins de 0 degre : Il fait froid ! Mettez votre manteau.
Entre 0 et 15 degres : Il fait frais. Pensez a prendre un pull.
Entre 16 et 23 degres : Beau temps.
Entre 24 et 32 degres : Il fait chaud, pensez a bien vous hydrater.
Plus de 32 degres : Il fait tres chaud ! Restez au frais.


Etape 4 : Temperatures a virgule flottante (Bonus)

Modifier get_random_temp() pour renvoyer un nombre a virgule flottante aleatoire afin d'obtenir des valeurs de temperature plus precises.


Etape 5 : Saisons mensuelles (Bonus)

Au lieu de generer directement une temperature aleatoire, demandez a l'utilisateur un mois (1-12) et determinez la saison en utilisant if/ elif conditions.
Modifier get_random_temp() pour obtenir des temperatures specifiques a chaque saison.


Resultat attendu :

The temperature right now is 32 degrees Celsius.
It's really hot! Stay cool.

```python
# Exercice 7 : Conseils sur la temperature
import random

def get_random_temp(saison=None):
    if saison == "ete":
        return round(random.uniform(20, 40), 1)
    elif saison == "hiver":
        return round(random.uniform(-10, 10), 1)
    elif saison == "printemps":
        return round(random.uniform(10, 22), 1)
    elif saison == "automne":
        return round(random.uniform(8, 20), 1)
    else:
        return round(random.uniform(-10, 40), 1)

def main():
    temp = get_random_temp()
    print(f"The temperature right now is {temp} degrees Celsius.")

    if temp < 0:
        print("Il fait froid ! Mettez votre manteau.")
    elif temp <= 15:
        print("Il fait frais. Pensez a prendre un pull.")
    elif temp <= 23:
        print("Beau temps.")
    elif temp <= 32:
        print("Il fait chaud, pensez a bien vous hydrater.")
    else:
        print("Il fait tres chaud ! Restez au frais.")

main()

# Bonus : avec saison
mois = int(input("Entrez le mois (1-12) : "))
if mois in [12, 1, 2]:
    saison = "hiver"
elif mois in [3, 4, 5]:
    saison = "printemps"
elif mois in [6, 7, 8]:
    saison = "ete"
else:
    saison = "automne"

temp_saison = get_random_temp(saison)
print(f"\nTemperature pour la saison {saison} : {temp_saison} degrees Celsius.")
```

 Exercice 8 : Garnitures de pizza
Principaux sujets en Python :

Boucles
Listes
formatage de chaines


Instructions:

Ecrivez une boucle qui demande a l'utilisateur de saisir les ingredients de la pizza un par un.

Arretez la boucle lorsque l'utilisateur saisit 'quit'.

Pour chaque garniture saisie, imprimez :
"Adding [topping] to your pizza."

Apres avoir quitte la boucle, affichez tous les ingredients et le prix total de la pizza.
Le prix de base est de 10 $, et chaque garniture ajoute 2,50 $.

```python
# Exercice 8 : Garnitures de pizza
toppings = []
prix_base = 10
prix_garniture = 2.5

while True:
    garniture = input("Entrez une garniture (ou 'quit' pour terminer) : ")
    if garniture.lower() == 'quit':
        break
    toppings.append(garniture)
    print(f"Adding {garniture} to your pizza.")

print("\n--- Recapitulatif de votre pizza ---")
print("Garnitures :", ", ".join(toppings))
total = prix_base + len(toppings) * prix_garniture
print(f"Prix total : {total} $")
```
