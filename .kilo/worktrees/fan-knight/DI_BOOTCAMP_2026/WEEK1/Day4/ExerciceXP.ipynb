```python
# Étape 1 : Créer la classe Chien (Dog)
class Dog:
    def __init__(self, name, age, weight):
        # Initialisation des trois attributs demandés
        self.name = name
        self.age = age
        self.weight = weight

    def bark(self):
        # Renvoie le message d'aboiement
        return f"{self.name} aboie !"

    def run_speed(self):
        # Calcule et renvoie la vitesse de course
        return (self.weight / self.age) * 10

    def fight(self, other_dog):
        # Calcul de la puissance pour le chien actuel (self)
        puissance_self = self.run_speed() * self.weight
        
        # Calcul de la puissance pour l'adversaire (other_dog)
        puissance_other = other_dog.run_speed() * other_dog.weight
        
        # Détermination du gagnant
        if puissance_self > puissance_other:
            return f" Combat : {self.name} a battu {other_dog.name} !"
        elif puissance_other > puissance_self:
            return f" Combat : {other_dog.name} a battu {self.name} !"
        else:
            return f" Combat : Égalité parfaite entre {self.name} et {other_dog.name} !"


# =====================================================================
# Étape 2 : Créer des instances de chien
# =====================================================================
# Création de 3 chiens avec des caractéristiques variées
dog1 = Dog(name="Rex", age=4, weight=25)       # Jeune et lourd
dog2 = Dog(name="Toby", age=2, weight=12)      # Très jeune et léger
dog3 = Dog(name="Max", age=8, weight=30)       # Âgé et très lourd


# =====================================================================
# Étape 3 : Méthodes de test sur les chiens
# =====================================================================
print("--- 1. Test des Aboiements (bark) ---")
print(dog1.bark())
print(dog2.bark())
print(dog3.bark())

print("\n--- 2. Test des Vitesses de course (run_speed) ---")
print(f"Vitesse de {dog1.name} : {dog1.run_speed():.2f} km/h")
print(f"Vitesse de {dog2.name} : {dog2.run_speed():.2f} km/h")
print(f"Vitesse de {dog3.name} : {dog3.run_speed():.2f} km/h")

print("\n--- 3. Test des Combats (fight) ---")
# Combat 1 : Rex contre Toby
print(dog1.fight(dog2))

# Combat 2 : Toby contre Max
print(dog2.fight(dog3))

# Combat 3 : Max contre Rex
print(dog3.fight(dog1))
```

Ajoutez une méthode appelée born(first_name, age):

N'oubliez pas de tester vos classes en créant une instance Family, en ajoutant des membres et en appelant chaque méthode pour voir le résultat attendu.



```python
# Étape 1 : Créer la classe Person
class Person:
    def __init__(self, first_name, age):
        self.first_name = first_name
        self.age = age
        self.last_name = ""  # Initialisé comme une chaîne vide

    def is_18(self):
        # Renvoie True si l'âge est supérieur ou égal à 18, sinon False
        return self.age >= 18


# Étape 2 : Créer la classe Family
class Family:
    def __init__(self, last_name):
        self.last_name = last_name
        self.members = []  # Liste vide pour stocker les objets Person

    def born(self, first_name, age):
        # Création d'un nouvel objet Person
        new_person = Person(first_name, age)
        # Attribution du nom de famille de la Family à la nouvelle Person
        new_person.last_name = self.last_name
        # Ajout de l'objet à la liste des membres
        self.members.append(new_person)
        print(f"👶 Bienvenue à {first_name} {self.last_name} dans la famille !")

    def check_majority(self, first_name):
        # Recherche de la personne dans la liste des membres
        personne_trouvee = None
        for member in self.members:
            if member.first_name == first_name:
                personne_trouvee = member
                break  # On arrête la recherche dès qu'on trouve le prénom

        # Si la personne a été trouvée dans la famille
        if personne_trouvee:
            # Appel de la méthode de l'objet Person pour vérifier sa majorité
            if personne_trouvee.is_18():
                print(f"➡️ {first_name} : You are over 18, your parents Jane and John accept that you will go out with your friends")
            else:
                print(f"➡️ {first_name} : Sorry, you are not allowed to go out with your friends.")
        else:
            print(f"❌ Aucune personne nommée {first_name} ne fait partie de cette famille.")

    def family_presentation(self):
        print(f"\n=== Présentation de la Famille {self.last_name.upper()} ===")
        for member in self.members:
            print(f"- {member.first_name}, {member.age} ans")


# =====================================================================
# Étape 3 : Test et vérification du programme
# =====================================================================

# 1. Création d'une instance de Family
ma_famille = Family("Smith")

# 2. Ajout de membres avec la méthode born()
print("--- Phase de naissances ---")
ma_famille.born("Alice", 22)  # Majeure
ma_famille.born("Bob", 14)    # Mineur
ma_famille.born("Charlie", 18) # Tout juste majeur

# 3. Affichage de la présentation familiale
ma_famille.family_presentation()

print("\n--- Phase de vérification des sorties (check_majority) ---")
# 4. Test des autorisations de sortie
ma_famille.check_majority("Alice")
ma_famille.check_majority("Bob")
ma_famille.check_majority("Charlie")

# Test bonus : recherche d'un prénom inexistant
ma_famille.check_majority("David")
```
