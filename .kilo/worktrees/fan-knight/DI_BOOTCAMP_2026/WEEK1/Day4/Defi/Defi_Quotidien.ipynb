Instructions : Système de pagination

#### Qu'est-ce que la pagination ?

En développement web, la pagination permet de diviser les longues listes en sections plus petites et plus faciles à gérer (pages), ce qui facilite la navigation dans des contenus tels que les résultats de recherche, les listes de produits ou les articles.

Voici un exemple visuel :

Page 1      Page 2      Page 3
[a, b, c]   [d, e, f]   [g, h, i]

Étape 4 : Implémenter les méthodes de navigation

Ces méthodes devraient faciliter la navigation entre les pages :

→ Accède à la page spécifiée (indexation à partir de 1).
→ Si page_numla page est hors limites, une exception est levée ValueError.

→ Permet d'accéder à la première page.

last_page()
→ Permet d'accéder à la dernière page.

→ Passe à la page suivante (si ce n'est pas déjà la dernière page).

previous_page()
→ Replonge d'une page (si la personne n'est pas déjà sur la première page).

#### Remarque :

Les pages sont indexées en interne à partir de 0, mais les entrées utilisateur doivent commencer à 1.
Toutes les méthodes de navigation (sauf go_to_page) doivent retourner une valeur selfpour permettre l'enchaînement des méthodes.

Bonus : améliorez votre code en modifiant l’instruction de retour de manière à pouvoir concaténer des méthodes comme ceci :

résultat : ['m', 'n', 'o', 'p']

```python
import math

# Étape 1 : Créer la classe Pagination
class Pagination:
    
    # Étape 2 : Implémenter la méthode __init__
    def __init__(self, items=None, page_size=10):
        # Si items est None, on initialise une liste vide
        self.items = items if items is not None else []
        
        # Conversion explicite en entier pour la sécurité du type
        self.page_size = int(page_size)
        
        # L'index de la page actuelle commence à 0 (interne)
        self.current_idx = 0
        
        # Calcul du nombre total de pages (minimum 1 page, même si vide)
        if len(self.items) == 0:
            self.total_pages = 1
        else:
            self.total_pages = math.ceil(len(self.items) / self.page_size)

    # Étape 3 : Mettre en œuvre la méthode get_visible_items()
    def get_visible_items(self):
        # Calcul des index de début et de fin pour le découpage de la liste
        debut = self.current_idx * self.page_size
        fin = debut + self.page_size
        return self.items[debut:fin]

    # Étape 4 : Implémenter les méthodes de navigation
    def go_to_page(self, page_num):
        # Conversion en entier de l'entrée utilisateur
        page_num = int(page_num)
        
        # Les entrées utilisateur commencent à 1. On vérifie les limites.
        if page_num < 1 or page_num > self.total_pages:
            raise ValueError(f"Erreur : La page {page_num} est hors limites. Total pages : {self.total_pages}.")
        
        # Passage à l'index interne (base 0)
        self.current_idx = page_num - 1
        return self  # Permet le chaînage

    def first_page(self):
        self.current_idx = 0
        return self  # Permet le chaînage

    def last_page(self):
        self.current_idx = self.total_pages - 1
        return self  # Permet le chaînage

    def next_page(self):
        # Avance si on n'est pas déjà sur la dernière page
        if self.current_idx < self.total_pages - 1:
            self.current_idx += 1
        return self  # Permet le chaînage

    def previous_page(self):
        # Recule si on n'est pas déjà sur la première page
        if self.current_idx > 0:
            self.current_idx -= 1
        return self  # Permet le chaînage

    # Étape 5 (Bonus) : Ajouter une méthode __str__() personnalisée
    def __str__(self):
         elements_visibles = self.get_visible_items()
         # Convertit chaque élément en chaîne et les joint avec un saut de ligne
         return "\n".join(str(item) for item in elements_visibles)


# =====================================================================
# Étape 6 : Test du code (Validation avec les cas de test)
# =====================================================================
print("--- TEST ÉTAPE 6 : Validation des consignes ---")

alphabetList = list("abcdefghijklmnopqrstuvwxyz")
p = Pagination(alphabetList, 4)

# Test Page 1
print("Page 1 :", p.get_visible_items())  # ['a', 'b', 'c', 'd']

# Test Page 2 via next_page()
p.next_page()
print("Page 2 :", p.get_visible_items())  # ['e', 'f', 'g', 'h']

# Test Dernière Page
p.last_page()
print("Dernière Page :", p.get_visible_items())  # ['y', 'z']

# Test de la méthode magique __str__ (Étape 5)
print("\n--- Rendu __str__() sur la dernière page ---")
print(str(p))

print("\n--- Test des exceptions (ValueError) ---")
# Test de gestion des erreurs (Hors limites supérieures)
try:
    p.go_to_page(10)
except ValueError as e:
    print(e)

# Test de gestion des erreurs (Hors limites inférieures)
try:
    p.go_to_page(0)
except ValueError as e:
    print(e)


print("\n" + "="*50 + "\n")


# =====================================================================
# TEST DU BONUS : Concaténation / Chaînage de méthodes
# =====================================================================
print("--- TEST DU BONUS : Chaînage de méthodes ---")

# Réinitialisation de la pagination sur la première page
p.first_page()

# Chaînage de plusieurs méthodes à la suite grâce au "return self"
resultat_chainage = p.next_page().next_page().next_page().get_visible_items()
print("Résultat du chaînage (3 fois next_page) :", resultat_chainage)
# Attendu : page index 3 (4ème page), soit ['m', 'n', 'o', 'p']
```
