Défi 1


Demandez à l'utilisateur un number et un length.

Créez un programme qui affiche une liste de multiples du nombre jusqu'à ce que la longueur de la liste atteigne la longueur spécifiée.

Exemples :

number: 7 - length 5 ➞ [7, 14, 21, 28, 35]

number: 12 - length 10 ➞ [12, 24, 36, 48, 60, 72, 84, 96, 108, 120]

number: 17 - length 6 ➞ [17, 34, 51, 68, 85, 102]

```python
# Demander les informations à l'utilisateur
nombre = int(input("Entrez un chiffre : "))
longueur = int(input("Entrez une longueur : "))

# Créer la liste des multiples
multiples = []

for i in range(1, longueur + 1):
    multiples.append(nombre * i)

# Afficher le résultat
print(multiples)
```

Défi 2


Écrivez un programme qui demande une chaîne de caractères à l'utilisateur et affiche une nouvelle chaîne sans les lettres consécutives identiques.

Exemples

```python
# Demander un mot à l'utilisateur
mot = input("Entrez un mot : ")

# Construire le nouveau mot
resultat = ""

for lettre in mot:
    if resultat == "" or lettre != resultat[-1]:
        resultat += lettre

# Afficher le résultat
print(resultat)
```
