Étape 6 : Mettre en œuvre la get_animal_typesméthode

Appelez la get_animal_typesméthode pour obtenir la liste des animaux.

Utilisez la mise en forme de chaînes de caractères pour créer le résultat.


Étape 8 : mettre à jour la méthode add_animal

À utiliser **kwargspour le transfert de plusieurs animaux. Les clés seront le nom de l'animal et les valeurs, la quantité.
Vous pouvez alors appeler la méthode de cette manière :macdonald.add_animal('cow'= 5, 'sheep' = 2, 'goat' = 12)


```python
class Farm:
    # Étape 1 & 2 : Créer la classe et implémenter la méthode __init__
    def __init__(self, farm_name):
        self.name = farm_name
        self.animals = {}  # Dictionnaire vide pour stocker {'animal': quantité}

    # Étape 3 & Étape 8 (Bonus) : Mettre en œuvre et mettre à jour la méthode add_animal
    def add_animal(self, animal_type=None, count=1, **kwargs):
        # Gestion du cas standard : macdonald.add_animal('cow', 5) ou macdonald.add_animal('sheep')
        if animal_type:
            if animal_type in self.animals:
                self.animals[animal_type] += count
            else:
                self.animals[animal_type] = count

        # Gestion du cas Bonus (**kwargs) : macdonald.add_animal(cow=5, sheep=2)
        # Note : en Python, les clés reçues dans kwargs ne prennent pas de guillemets lors de l'appel
        for anim, qty in kwargs.items():
            if anim in self.animals:
                self.animals[anim] += qty
            else:
                self.animals[anim] = qty

    # Étape 4 : Mettre en œuvre la méthode get_info
    def get_info(self):
        # En-tête demandé par l'output d'exemple
        info_str = f"{self.name}'s farm\n"
        
        # Parcours du dictionnaire pour aligner les noms et le nombre d'animaux
        for animal, count in self.animals.items():
            info_str += f"{animal} : {count}\n"
            
        # Ajout de la phrase finale avec un espacement (indentation) conforme à l'exemple
        info_str += "    E-I-E-I-0!"
        return info_str

    # Étape 6 (Bonus) : Mettre en œuvre la méthode get_animal_types
    def get_animal_types(self):
        # Renvoie la liste triée des clés du dictionnaire
        return sorted(list(self.animals.keys()))

    # Étape 7 (Bonus) : Mettre en œuvre la méthode get_short_info
    def get_short_info(self):
        types_tries = self.get_animal_types()
        animaux_pluriel = []

        # Construction des noms au pluriel si la quantité > 1
        for animal in types_tries:
            if self.animals[animal] > 1:
                # Gestion simple du pluriel en français (ajout d'un 's')
                animaux_pluriel.append(f"{animal}s")
            else:
                animaux_pluriel.append(animal)

        # Formatage de la liste de chaînes (ex: "vaches, chèvres et moutons")
        if len(animaux_pluriel) > 1:
            # On joint tous les éléments sauf le dernier par des virgules, puis on ajoute le dernier avec "et"
            liste_formatee = ", ".join(animaux_pluriel[:-1]) + f" et {animaux_pluriel[-1]}"
        elif len(animaux_pluriel) == 1:
            liste_formatee = animaux_pluriel[0]
        else:
            liste_formatee = "aucun animal"

        return f"La ferme de {self.name} possède des {liste_formatee}."


# =====================================================================
# Étape 5 : Test du code (Validation de l'exemple)
# =====================================================================
print("--- TEST ÉTAPES 1 À 5 (Sortie standard attendue) ---")
macdonald = Farm("McDonald")
macdonald.add_animal('cow', 5)
macdonald.add_animal('sheep')
macdonald.add_animal('sheep')
macdonald.add_animal('goat', 12)

print(macdonald.get_info())


print("\n" + "="*50 + "\n")


# =====================================================================
# TEST DES BONUS (Étapes 6, 7 et 8)
# =====================================================================
print("--- TEST DES ÉTAPES BONUS (6, 7 et 8) ---")

# Création d'une nouvelle ferme pour tester l'extension
ferme_bonus = Farm("Avenir")

# Étape 8 : Ajout multiple en une seule ligne via **kwargs
# Attention : lors de l'appel avec des mots-clés (=), on ne met pas de guillemets sur les clés à gauche
ferme_bonus.add_animal(vache=5, mouton=1, cheval=3)
print("\nContenu de la ferme Bonus après kwargs :")
print(ferme_bonus.get_info())

# Étape 6 : Récupération des types triés alphabétiquement
print("\nListe triée des types d'animaux :")
print(ferme_bonus.get_animal_types())

# Étape 7 : Affichage des informations courtes avec gestion du pluriel
print("\nPrésentation courte de la ferme :")
print(ferme_bonus.get_short_info())
```
