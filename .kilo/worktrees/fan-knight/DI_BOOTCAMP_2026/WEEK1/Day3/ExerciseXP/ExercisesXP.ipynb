Utilisez la Catclasse fournie pour créer trois objets chat. Ensuite, créez une fonction pour trouver le chat le plus âgé et afficher ses détails.



Étape 1 : Créer des objets Chat

Utilisez cette Catclasse pour créer trois objets chat avec des noms et des âges différents.


Étape 2 : Créer une fonction pour trouver le chat le plus vieux

Créez une fonction qui prend les trois objets chat en entrée.
Dans cette fonction, comparez l'âge des chats pour trouver le plus âgé.
Renvoie l'objet chat le plus ancien.


Étape 3 : Imprimer les informations du chat le plus âgé

Appelez la fonction pour obtenir le chat le plus âgé.
Afficher une chaîne formatée : « Le chat le plus âgé est <cat_name>, et a <cat_age>ans. »
Remplacez <cat_name>et <cat_age>par le nom et l'âge du chat le plus âgé.


Exemple:

```python
class Cat:
    def __init__(self, cat_name, cat_age):
        self.name = cat_name
        self.age = cat_age

# Étape 1 : Créer trois objets Chat (Instanciation)
cat1 = Cat("Minou", 5)
cat2 = Cat("Félix", 12)
cat3 = Cat("Choupette", 8)

# Étape 2 : Créer une fonction pour trouver le chat le plus vieux
def find_oldest_cat(c1, c2, c3):
    # Utilisation de la fonction max() combinée avec une clé (l'âge)
    # ou une structure de comparaison classique.
    # Ici, max() compare l'attribut 'age' de chaque objet chat fourni.
    oldest = max(c1, c2, c3, key=lambda cat: cat.age)
    return oldest

# Étape 3 : Imprimer les informations du chat le plus âgé
oldest_cat = find_oldest_cat(cat1, cat2, cat3)

# Affichage avec une chaîne de caractères formatée (f-string)
print(f"Le chat le plus âgé est {oldest_cat.name}, et a {oldest_cat.age} ans.")
```

Instructions:

Créez une Dogclasse avec des méthodes pour aboyer et sauter. Instanciez des objets chien, appelez leurs méthodes et comparez leurs tailles.



Étape 1 : Créer la classe Chien

Créez une classe appelée Dog.
Dans cette __init__méthode, prenez nameet heightcomme paramètres et créez les attributs correspondants.
Créez une bark()méthode qui affiche « <dog_name>fait ouaf ! ».
Créez une jump()méthode qui imprime « <dog_name>saute <x>cm de haut ! », où xest height * 2.


Étape 2 : Créer des objets Chien

Créez davids_dogdes sarahs_dogobjets avec leurs noms et hauteurs respectifs.


Étape 3 : Imprimer les informations sur le chien et les méthodes d’appel

Inscrivez le nom et la taille de chaque chien.
Appelez les méthodes bark()et jump()pour chaque chien.


Étape 4 : Comparer la taille des chiens

#### Réponse

```python
# Étape 1 : Créer la classe Chien
class Dog:
    def __init__(self, name, height):
        # Initialisation des attributs
        self.name = name
        self.height = height  # la hauteur/taille en cm

    def bark(self):
        # Méthode pour aboyer
        print(f"{self.name} fait ouaf !")

    def jump(self):
        # Méthode pour sauter (x vaut la hauteur multipliée par 2)
        x = self.height * 2
        print(f"{self.name} saute {x} cm de haut !")


# Étape 2 : Créer des objets Chien (Instanciation)
# Remplaçons par des valeurs d'exemple (ex: David a un Rex de 50cm, Sarah a un Mittens de 30cm)
davids_dog = Dog(name="Rex", height=50)
sarahs_dog = Dog(name="Mittens", height=30)


# Étape 3 : Imprimer les informations sur le chien et appeler les méthodes
print("--- Informations sur le chien de David ---")
print(f"Nom : {davids_dog.name} | Taille : {davids_dog.height} cm")
davids_dog.bark()
davids_dog.jump()

print("\n--- Informations sur le chien de Sarah ---")
print(f"Nom : {sarahs_dog.name} | Taille : {sarahs_dog.height} cm")
sarahs_dog.bark()
sarahs_dog.jump()


print("\n" + "="*40 + "\n")


# Étape 4 : Comparer la taille des chiens
print("--- Comparaison de la taille des chiens ---")
if davids_dog.height > sarahs_dog.height:
    print(f"Le plus grand chien est {davids_dog.name}, le chien de David.")
elif sarahs_dog.height > davids_dog.height:
    print(f"Le plus grand chien est {sarahs_dog.name}, le chien de Sarah.")
else:
    print(f"{davids_dog.name} et {sarahs_dog.name} ont exactement la même taille !")
```

Instructions:

Créez une Songclasse avec une méthode permettant d'imprimer les paroles d'une chanson ligne par ligne.



Étape 1 : Créer la classe de chansons

Créez une classe appelée Song.
Dans la __init__méthode, prenez lyrics(une liste) comme paramètre et créez un attribut correspondant.
Créez une sing_me_a_song()méthode qui affiche chaque élément de la lyricsliste sur une nouvelle ligne.


Exemple:

stairway = Song(["There’s a lady who's sure", "all that glitters is gold", "and she’s buying a stairway to heaven"])

stairway.sing_me_a_song()

Résultat : Il y a une dame qui est persuadée que tout ce qui brille n'est pas or et elle achète un escalier vers le paradis

```python
# Étape 1 : Créer la classe de chansons (Song)
class Song:
    def __init__(self, lyrics):
        # Initialisation de l'attribut avec la liste reçue en paramètre
        self.lyrics = lyrics

    def sing_me_a_song(self):
        # Utilisation d'une boucle 'for' pour parcourir la liste ligne par ligne
        for line in self.lyrics:
            print(line)


# Étape 2 : Instanciation et test avec l'exemple fourni
stairway = Song([
    "There’s a lady who's sure", 
    "all that glitters is gold", 
    "and she’s buying a stairway to heaven"
])

# Appel de la méthode pour chanter la chanson
print("--- Rendu de la chanson ---")
stairway.sing_me_a_song()
```

Bonus : Modifiez la add_animal()méthode pour éviter *argsde la répéter à chaque fois pour un nouvel animal ; vous pouvez désormais passer plusieurs noms d’animaux séparés par une virgule.

```python
class Zoo:
    # Étape 1 : Définir la Zoo classe
    def __init__(self, zoo_name):
        self.name = zoo_name
        self.animals = []
        # Attribut interne pour stocker le dictionnaire des groupes triés
        self.grouped_animals = {}

    # BONUS : Utilisation de *new_animals (*args) pour accepter plusieurs animaux
    def add_animal(self, *new_animals):
        for animal in new_animals:
            if animal not in self.animals:
                self.animals.append(animal)
                print(f"Ajouté : {animal}")
            else:
                print(f"L'animal {animal} est déjà présent dans le zoo.")

    def get_animals(self):
        print(f"\n--- Liste des animaux dans {self.name} ---")
        if not self.animals:
            print("Le zoo est actuellement vide.")
        else:
            print(", ".join(self.animals))

    def sell_animal(self, animal_sold):
        if animal_sold in self.animals:
            self.animals.remove(animal_sold)
            print(f"\nVendu : {animal_sold} a quitté le zoo.")
        else:
            print(f"\nImpossible de vendre : {animal_sold} ne fait pas partie du zoo.")

    def sort_animals(self):
        # 1. Trier la liste principale par ordre alphabétique
        self.animals.sort()
        
        # 2. Réinitialiser le dictionnaire de regroupement
        self.grouped_animals = {}
        
        # 3. Regrouper par la première lettre
        for animal in self.animals:
            premiere_lettre = animal[0].upper() # On s'assure d'avoir une majuscule
            
            # Si la lettre n'est pas encore une clé, on crée une liste vide
            if premiere_lettre not in self.grouped_animals:
                self.grouped_animals[premiere_lettre] = []
                
            # On ajoute l'animal à la liste correspondant à sa lettre
            self.grouped_animals[premiere_lettre].append(animal)
            
        return self.grouped_animals

    def get_groups(self):
        print(f"\n--- Groupes d'animaux triés par lettre ({self.name}) ---")
        if not self.grouped_animals:
            print("Veuillez d'abord exécuter sort_animals() pour générer les groupes.")
        else:
            for lettre, liste_animaux in self.grouped_animals.items():
                print(f"{lettre}: {liste_animaux}")


# =====================================================================
# Étape 2 : Créer un objet Zoo
# =====================================================================
brooklyn_safari = Zoo("Brooklyn Safari")


# =====================================================================
# Étape 3 : Appeler les méthodes du zoo pour tester
# =====================================================================

# 1. Ajout d'animaux (Grâce au bonus, on en passe plusieurs d'un coup)
print("--- Phase d'ajout ---")
brooklyn_safari.add_animal("Giraffe", "Bear", "Baboon", "Zebra", "Cat", "Cougar", "Lion")
# Test de doublon
brooklyn_safari.add_animal("Giraffe") 

# 2. Affichage initial
brooklyn_safari.get_animals()

# 3. Vendre un animal et ré-afficher
brooklyn_safari.sell_animal("Bear")
brooklyn_safari.get_animals()

# 4. Tri et regroupement
brooklyn_safari.sort_animals()

# 5. Affichage des groupes
brooklyn_safari.get_groups()
```
